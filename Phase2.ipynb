```python
import pandas as pd

file_path = "./-IT326-Project/Dataset/Raw_Dataset.csv"
df = pd.read_csv(file_path)

df.head()
```

```python
# Missing-value analysis
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_analysis = pd.DataFrame({
    'Missing Values': missing_count,
    'Percentage (%)': missing_percentage
})

missing_analysis
```

```python
# Examine columns with missing values
missing_columns = df.columns[df.isnull().any()]

df[missing_columns].describe()
```

```python
# Check skewness of columns with missing values
df[missing_columns].skew()
```

```python
# Preprocessing Technique #1: Mean Imputation

df_preprocessed = df.copy()

for column in missing_columns:
    df_preprocessed[column] = df_preprocessed[column].fillna(
        df_preprocessed[column].mean()
    )

# Check missing values after preprocessing
df_preprocessed[missing_columns].isnull().sum()
```

```python
# Compare missing values before and after preprocessing

comparison = pd.DataFrame({
    'Before': df[missing_columns].isnull().sum(),
    'After': df_preprocessed[missing_columns].isnull().sum()
})

comparison
```

```python
# remove the user identifier
df_preprocessed = df_preprocessed.drop(columns=['User_ID'])
```

**User id is only an identifier and does not provide useful informatioon for analysis.**

```python
# identify numerical attributes before converting categorical attributes
numeric_columns = df_preprocessed.select_dtypes(include='number').columns
```

```python
# Preprocessing Technique #2: Normalization through Min-Max Scaling

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df_preprocessed[numeric_columns] = scaler.fit_transform(df_preprocessed[numeric_columns])
```

**numerical attributes have different ranges, so normalization was applied to put them on the same scale**

```python
df_preprocessed.head()
```

**All numerical attributes were scaled to a range of 0–1**

```python
# Preprocessing Technique #3: Data Conversion through Label Encoding

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df_preprocessed['Gender'] = le.fit_transform(df_preprocessed['Gender'])
df_preprocessed['Physical_Activity_Level'] = le.fit_transform(df_preprocessed['Physical_Activity_Level'])
df_preprocessed['Diet_Type'] = le.fit_transform(df_preprocessed['Diet_Type'])
df_preprocessed['Daily_Stress_Level'] = le.fit_transform(df_preprocessed['Daily_Stress_Level'])
```

**Categorical attributes were represented as text and needed to be converted into numerical values for the data mining algorithms.**

```python
df_preprocessed.head()
```

**Gender, Physical Activity Level, Diet Type, and Daily Stress Level were converted into numerical labels.**

```python
# Comparison of raw and preprocessed datasets

comparison = pd.DataFrame(
    {
    '': ['Number of Records',
         'Number of Attributes',
         'Missing Values',
         'User_ID', 'Numerical Attributes', 'Categorical Attributes'],

    'Raw Dataset': [df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        'Present',
        'Original scale',
        'Text'],
    'Preprocessed Dataset':[
        df_preprocessed.shape[0],
        df_preprocessed.shape[1],
        df_preprocessed.isnull().sum().sum(),
        'Removed',
        'Normalized (0–1)',
        'Label encoded']
    })

display(comparison)
```

```python
print("Raw Dataset")
display(df.head())

print("Preprocessed Dataset")
display(df_preprocessed.head())
```

### Overall:
Missing values were handled using mean imputation, reducing the total number of missing values from 825 to 0. The numerical attributes were then normalized to a range of 0 to 1, while categorical attributes were converted into numerical values. The User_ID attribute was also removed because it is only an identifier. The resulting dataset contains 10,000 records and 15 attributes and was saved separately as Preprocessed_dataset.csv.

```python
# Save the preprocessed dataset separately
output_path = "./-IT326-Project/Dataset/Preprocessed_dataset.csv"

df_preprocessed.to_csv(output_path, index=False)

print("Preprocessed dataset saved successfully.")
```
